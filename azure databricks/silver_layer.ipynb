```python
from pyspark.sql.functions import *
```

```python
%sql
CREATE EXTERNAL LOCATION IF NOT EXISTS silver_raw
URL 'abfss://silver@adventworksdatalake.dfs.core.windows.net/'
WITH (STORAGE CREDENTIAL azure_workspace_inst);
```

```python
%sql SHOW EXTERNAL LOCATIONS;
```

```python
path = "abfss://bronze@adventworksdatalake.dfs.core.windows.net/calendar/calendar.csv"
calendar = (spark.read
      .format("csv")
      .option("header", True)
      .option("inferSchema", True)
      .load(path))

categories = (spark.read
      .format("csv")
      .option("header", True)
      .option("inferSchema", True)
      .load("abfss://bronze@adventworksdatalake.dfs.core.windows.net/category/product_categories.csv"))

customers = (spark.read
      .format("csv")
      .option("header", True)
      .option("inferSchema", True)
      .load("abfss://bronze@adventworksdatalake.dfs.core.windows.net/customers/customers.csv"))

products = (spark.read
      .format("csv")
      .option("header", True)
      .option("inferSchema", True)
      .load("abfss://bronze@adventworksdatalake.dfs.core.windows.net/products/products.csv"))


returns = (spark.read
      .format("csv")
      .option("header", True)
      .option("inferSchema", True)
      .load("abfss://bronze@adventworksdatalake.dfs.core.windows.net/returns/returns.csv"))



sales_2015 = (spark.read
      .format("csv")
      .option("header", True)
      .option("inferSchema", True)
      .load("abfss://bronze@adventworksdatalake.dfs.core.windows.net/sales_2015/sales_2015.csv"))





sales_2016 = (spark.read
      .format("csv")
      .option("header", True)
      .option("inferSchema", True)
      .load("abfss://bronze@adventworksdatalake.dfs.core.windows.net/sales_2016/sales_2016.csv"))


sales_2017 = (spark.read
      .format("csv")
      .option("header", True)
      .option("inferSchema", True)
      .load("abfss://bronze@adventworksdatalake.dfs.core.windows.net/sales_2017/sales_2017.csv"))

subcategory = (spark.read
      .format("csv")
      .option("header", True)
      .option("inferSchema", True)
      .load("abfss://bronze@adventworksdatalake.dfs.core.windows.net/subcategory/product_subcategories.csv"))


territories = (spark.read
      .format("csv")
      .option("header", True)
      .option("inferSchema", True)
      .load("abfss://bronze@adventworksdatalake.dfs.core.windows.net/territories/territories.csv"))






```

Transformations


```python
calendar_Df= calendar.withColumn("Year", year(col("Date"))).withColumn("Month", month(col("Date")))
```

```python
calendar_Df.write.format('parquet')\
            .mode('append')\
            .option("path","abfss://silver@adventworksdatalake.dfs.core.windows.net/calendar/")\
            .save()
```

```python
customers.withColumn("fullName",concat(col('Prefix'),lit(' '),col('FirstName'),lit(' '),col('LastName'))).display()
customers.write.format('parquet')\
            .mode('append')\
            .option("path","abfss://silver@adventworksdatalake.dfs.core.windows.net/customers/")\
            .save()
```

```python
products= products.withColumn('ProductSKU',split(col('ProductSKU'),'-')[0])\
                .withColumn('ProductName',split(col('ProductName'),' ')[0])

products.write.format('parquet')\
            .mode('append')\
            .option("path","abfss://silver@adventworksdatalake.dfs.core.windows.net/products/")\
            .save()
```

```python
returns.write.format('parquet')\
            .mode('append')\
            .option("path","abfss://silver@adventworksdatalake.dfs.core.windows.net/returns/")\
            .save()
```

```python
territories.write.format('parquet')\
            .mode('append')\
            .option("path","abfss://silver@adventworksdatalake.dfs.core.windows.net/territories/")\
            .save()

```

```python
subcategory.write.format('parquet')\
            .mode('append')\
            .option("path","abfss://silver@adventworksdatalake.dfs.core.windows.net/subcategory/")\
            .save()
```

```python
df = sales_2017.unionByName(sales_2015).unionByName(sales_2016)
```

```python
df.write.format('parquet')\
            .mode('append')\
            .option("path","abfss://silver@adventworksdatalake.dfs.core.windows.net/sales/")\
            .save()
```
